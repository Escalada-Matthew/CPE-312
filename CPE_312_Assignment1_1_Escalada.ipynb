1. Overfitting and Underfitting are the two failure modes when training a model. Overfitting is when the the model is too complex. It focuses on memorizing training data instead of learning general patterns. It is unable to perform well when given new or test data. Underfitting is when the model is too simple that it fails to capture real patterns. It fails in both training and testing data.

An example of **Overfitting** is when creating a road damage detection model. It is able to correctly classify the training images but unable to classify new images of roads.

An example of **Underfitting** is when you create a model that predicts earthquake occurences but failing to implement features like location and magnitude, leading to poor predictions.

Cross fold validation is a technique where the dataset is split into k subsets or folds. the model is trained k-1 folds and tested on the remaining folds. The process will repeat depending on how many is k and the average performance is taken. It is useful in detecting overfitting by showing if performance will vary drastically across all the folds.

```python
#2 Demonstrate

from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
iris = load_iris()
X = iris.data
y = iris.target

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

print("Training set size:", X_train.shape[0])
print("Testing set size:", X_test.shape[0])
```

```python
#3 Demonstrate

from sklearn.metrics import precision_score, recall_score, f1_score

TP = 600
FP = 100
FN = 300
TN = 9000

y_true = ([1] * TP) + ([1] * FN) + ([0] * FP) + ([0] * TN)
y_pred = ([1] * TP) + ([0] * FN) + ([1] * FP) + ([0] * TN)

precision = precision_score(y_true, y_pred)
recall = recall_score(y_true, y_pred)
f1 = f1_score(y_true, y_pred)

print("Precision:", precision)
print("Recall:", recall)
print("F1-score:", f1)
```

 How can we derive accuracy of the model based on the previously mentioned metrics?

 The derivation of accuracy uses a formula where you use the relationships between TP, FP, TN, FN. The formula is TP+TN/TP, TN, FP, FN.
